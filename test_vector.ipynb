```python
import requests
import numpy as np

def generate_embeddings(text):
    url = "https://api.euron.one/api/v1/euri/embeddings"
    headers = {
        "Content-Type": "application/json",
        "Authorization": "Bearer euri-8d972efcbef7e46216237b101b200f7330bfb4666f0c97215c6f51a300780155"
    }
    payload = {
        "input": text,
        "model": "text-embedding-3-small"
    }

    response = requests.post(url, headers=headers, json=payload)
    data = response.json()
    
    embedding = np.array(data['data'][0]['embedding'])
    
    return embedding

text = "My name is amarendra kumar kushwaha. I am a software engineer. I love to code."

embedding = generate_embeddings(text)
```
